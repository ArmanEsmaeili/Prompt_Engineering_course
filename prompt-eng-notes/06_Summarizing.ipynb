````python
from google.colab import userdata
from together import Together

# Initialize the Together client with the API key
client = Together(api_key=userdata.get('TOGETHER_API_KEY'))
````

````python
def get_completion(prompt, model="meta-llama/Llama-2-13b-chat-hf"):
    response = client.completions.create(
        model=model,
        prompt=prompt,
        temperature=0.0,  # Adjusting the temperature for better randomness control
    )
    return response.choices[0].text
````

````python
prod_review = """
Got this panda plush toy for my daughter's birthday, \
who loves it and takes it everywhere. It's soft and \
super cute, and its face has a friendly look. It's \
a bit small for what I paid though. I think there \
might be other options that are bigger for the \
same price. It arrived a day earlier than expected, \
so I got to play with it myself before I gave it \
to her.
"""

# این عروسک پاندا را برای تولد دخترم خریدم که خیلی دوستش دارد و همه جا با خودش می‌برد.
# خیلی نرم و فوق‌العاده بامزه است و چهره‌ای دوستانه دارد.
# فقط کمی کوچکتر از چیزی است که بابتش پول پرداخت کردم.
# فکر می‌کنم گزینه‌های دیگری با همان قیمت و اندازه بزرگتر وجود داشته باشند.
# یک روز زودتر از انتظار رسید، بنابراین قبل از اینکه به او بدهم، خودم هم با آن بازی کردم.
````

````python
prompt = f"""
You are an AI language model designed to summarize text.
Please provide a concise summary of the following product
review from an ecommerce site in at most 30 words .

Review: ```{prod_review}```
"""

response = get_completion(prompt)
print(response)
````
