````python
from google.colab import userdata
from together import Together

# Initialize the Together client with the API key
client = Together(api_key=userdata.get('TOGETHER_API_KEY'))
````

````python
def get_completion(prompt, model="meta-llama/Llama-3-70b-chat-hf"):
    response = client.completions.create(
        model=model,
        prompt=prompt,
        temperature=0.0,  # Adjusting the temperature for better randomness control
    )
    return response.choices[0].text
````

````python
prompt = """
Translate the following English text just one time to Spanish: \
Hi, I would like to order a blender
"""
response = get_completion(prompt)
print(response)
````

````python
user_messages = [
  "La performance du système est plus lente que d'habitude.",  # System performance is slower than normal
  "Mi monitor tiene píxeles que no se iluminan.",              # My monitor has pixels that are not lighting
  "Il mio mouse non funziona",                                 # My mouse is not working
  "Mój klawisz Ctrl jest zepsuty",                             # My keyboard has a broken control key
  "我的屏幕在闪烁"                                              # My screen is flashing
]
````

````python
for issue in user_messages:
    # Create a clear and refined prompt
    prompt = f"""
    Task:
    1. Identify the language of the given text.
    2. Translate the text into English.
    3. Translate the text into Persian.

    Input Text: '''{issue}'''

    Please provide the output in the following format:
    Original: {issue}
    Language: [Detected language]
    English: [English translation]
    Persian: [Persian translation]
    """

    # Get the response from the model
    response = get_completion(prompt)

    # Parse and filter the response to extract the relevant parts
    relevant_keys = ["Original:", "Language:", "English:", "Persian:"]
    filtered_output = '\n'.join(
        line for line in response.strip().split('\n')
        if any(key in line for key in relevant_keys)
    )


    # Print the filtered output with a newline for better readability
    print(filtered_output + "\n")
````
