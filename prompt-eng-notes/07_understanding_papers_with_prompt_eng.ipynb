````python
prompt = f"""
what is the idea of following paper

Paper: ```{paper_contents}```
"""

response = get_completion(prompt)
print(response)
````

````python
# Define the questions as before
Q1 = """What problem is the paper trying to solve?"""
Q2 = """Why is the problem interesting?"""
Q3 = """What is the primary contribution?"""
Q4 = """How did they do it?"""
Q5 = """What are the key take-aways?"""

Questions = [Q1, Q2, Q3, Q4, Q5]

# Assuming the function to call Llama-3.2-11B-Vision-Instruct-Turbo is `llama_completion`

for i in range(len(Questions)):
    prompt = f"""
    You are a highly skilled research assistant tasked with analyzing a research paper and answering questions based on its content.

    Question:
    {Questions[i]}

    Research Paper:
    {paper_contents}
    """

    # Get response from the model
    response = get_completion(prompt)

    from IPython.display import Markdown
    # Print output in a well-structured format
    display(Markdown(f"### Question {i+1}: {Questions[i]}\n"))
    display(Markdown(f"**Answer:** {response}\n"))
    display(Markdown("****\n"))
````
