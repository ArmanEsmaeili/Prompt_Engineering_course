````python
from google.colab import userdata
from together import Together

# Initialize the Together client with the API key
client = Together(api_key=userdata.get('TOGETHER_API_KEY'))
````

````python
def get_completion(prompt, model="meta-llama/Llama-2-13b-chat-hf"):
    response = client.completions.create(
        model=model,
        prompt=prompt,
        temperature=0.0,  # Adjusting the temperature for better randomness control
    )
    return response.choices[0].text
````

````python
story = """
In a recent survey conducted by the government,
public sector employees were asked to rate their level
of satisfaction with the department they work at.
The results revealed that NASA was the most popular
department with a satisfaction rating of 95%.

One NASA employee, John Smith, commented on the findings,
stating, "I'm not surprised that NASA came out on top.
It's a great place to work with amazing people and
incredible opportunities. I'm proud to be a part of
such an innovative organization."

The results were also welcomed by NASA's management team,
with Director Tom Johnson stating, "We are thrilled to
hear that our employees are satisfied with their work at NASA.
We have a talented and dedicated team who work tirelessly
to achieve our goals, and it's fantastic to see that their
hard work is paying off."

The survey also revealed that the
Social Security Administration had the lowest satisfaction
rating, with only 45% of employees indicating they were
satisfied with their job. The government has pledged to
address the concerns raised by employees in the survey and
work towards improving job satisfaction across all departments.
"""
# در یک نظرسنجی اخیر که توسط دولت انجام شد،
# از کارکنان بخش عمومی خواسته شد تا سطح رضایت خود از دپارتمانی که در آن کار می‌کنند را ارزیابی کنند.
# نتایج نشان داد که ناسا محبوب‌ترین دپارتمان است و امتیاز رضایتی معادل ۹۵٪ دارد.

# یکی از کارکنان ناسا، جان اسمیت، در مورد این یافته‌ها اظهار نظر کرد و گفت:
# "من تعجب نمی‌کنم که ناسا در صدر قرار گرفته است.
# اینجا مکانی فوق‌العاده برای کار کردن با افراد شگفت‌انگیز و فرصت‌های بی‌نظیر است.
# من افتخار می‌کنم که بخشی از چنین سازمان نوآورانه‌ای هستم."

# این نتایج همچنین توسط تیم مدیریت ناسا مورد استقبال قرار گرفت،
# و مدیر ناسا، تام جانسون، اظهار داشت:
# "ما از شنیدن این خبر که کارکنان ما از کارشان در ناسا رضایت دارند، بسیار خوشحالیم.
# ما تیمی بااستعداد و متعهد داریم که بدون خستگی برای دستیابی به اهدافمان تلاش می‌کنند،
# و فوق‌العاده است که می‌بینیم تلاش‌های سخت آن‌ها به نتیجه می‌رسد."

# این نظرسنجی همچنین نشان داد که اداره تأمین اجتماعی کمترین امتیاز رضایت را داشته است،
# به طوری که تنها ۴۵٪ از کارکنان اعلام کردند که از شغل خود رضایت دارند.
# دولت متعهد شده است که به نگرانی‌های مطرح‌شده توسط کارکنان در این نظرسنجی رسیدگی کند
# و برای بهبود رضایت شغلی در تمام دپارتمان‌ها تلاش کند.
````

متن بالا را در نظر بگیرید:

1. ابتدا کلمات کلیدی که نشان‌دهنده موضوع متن هستند را استخراج کنید.
2. سپس، یک لیست از کلمات کلیدی ایجاد کنید که شامل کلماتی باشد که خودتان اضافه کرده‌اید (برخی از آنها در متن موجود نیستند).
3. از مدل بخواهید متن را بررسی کند و به ازای هر کلمه پاسخ دهد که آیا آن کلمه در متن وجود دارد یا خیر.
4. خروجی به صورت عددی باشد:
   - "1" برای وجود کلمه در متن.
   - "0" برای عدم وجود کلمه در متن.
